```python
import pandas as pd

# 1. Create a dataset with intentional duplicates
data = {
    'Order_ID': ['ORD-01', 'ORD-02', 'ORD-03', 'ORD-01', 'ORD-04', 'ORD-05', 'ORD-05'],
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Laptop', 'Monitor', 'Desk', 'Chair'],
    'Sales': [1200, 25, 100, 1200, 300, 450, 450]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
```

```python

full_duplicates = df.duplicated().sum()
print(f"Full Row Duplicates Found: {full_duplicates}")


key_duplicates = df.duplicated(subset=['Order_ID']).sum()
print(f"Key Column (Order_ID) Duplicates Found: {key_duplicates}")
```

```python

df_cleaned = df.drop_duplicates()

print("\nCleaned Data (Full Duplicates Removed):")
print(df_cleaned)
```
